```python
import os
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR_2 = "data/fc27_data"
print(os.listdir(DATA_DIR_2))

df2 = pd.read_csv(f"{DATA_DIR_2}/players.csv")
print(df2.columns.tolist())

print(df2["overall_rating"].describe())
print(df2["position"].unique())

# Dataset 2 - FC27 (football)
total_cells_2 = df2.size
missing_cells_2 = df2.isnull().sum().sum()
pct_missing_2 = (missing_cells_2 / total_cells_2) * 100

print(f"Dataset 2 - Total cells: {total_cells_2}")
print(f"Dataset 2 - Missing cells: {missing_cells_2}")
print(f"Dataset 2 - Overall missing: {pct_missing_2:.2f}%")
print(f"Duplicate rows: {df2.duplicated().sum()}")
```
