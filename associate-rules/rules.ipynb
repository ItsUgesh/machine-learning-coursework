```python
sessions = [
    {"Videos", "Quiz", "Flashcards"},  # 1
    {"Videos", "Quiz"},                # 2
    {"Videos", "Notes"},               # 3
    {"Videos", "Quiz", "Notes"},       # 4
    {"Quiz", "Flashcards"},            # 5
    {"Notes", "Discussion"},           # 6
    {"Videos", "Discussion"},          # 7
    {"Quiz", "Notes"},                 # 8
    {"Videos", "Quiz", "Flashcards"},  # 9
    {"Notes", "Discussion"},           # 10
    {"Videos", "Quiz"},                # 11
    {"Quiz", "Flashcards"},            # 12
]


def metrics(a, b):
    """Return support, confidence and lift for the rule a -> b."""
    total = len(sessions)
    count_a = sum(a in s for s in sessions)
    count_b = sum(b in s for s in sessions)
    count_ab = sum(a in s and b in s for s in sessions)

    support = count_ab / total
    confidence = count_ab / count_a
    lift = confidence / (count_b / total)
    return support, confidence, lift
```

```python
rules = [
    ("Quiz", "Flashcards"),
    ("Notes", "Discussion"),
    ("Flashcards", "Quiz"),  # reversed strongest rule
]

for a, b in rules:
    s, c, l = metrics(a, b)
    print(f"{a} -> {b}: support={s:.3f}, confidence={c:.2f}, lift={l:.2f}")
```

```python
import pandas as pd

final_rules = [("Rule 1", "Quiz", "Flashcards"), ("Rule 2", "Notes", "Discussion")]

rows = []
for name, a, b in final_rules:
    s, c, l = metrics(a, b)
    rows.append({
        "Rule": f"{name}: {a} -> {b}",
        "Support": round(s, 3),
        "Confidence": round(c, 2),
        "Lift": round(l, 2),
    })

final_table = pd.DataFrame(rows)
print(final_table.to_string(index=False))

s, c, l = metrics("Quiz", "Flashcards")
print()
print(
    f"Our recommended rule is Quiz -> Flashcards because it has a support of {s:.3f}, "
    f"confidence of {c:.2f} and lift of {l:.1f}. We think this relationship is useful "
    "because it covers a third of all sessions, and every Flashcards session also used "
    "Quiz, so a Flashcards shortcut next to Quiz would reach many students."
)
```
