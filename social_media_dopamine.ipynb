```python
import os
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = "data/dopamine_data"
print(os.listdir(DATA_DIR))

df1 = pd.read_csv(f"{DATA_DIR}/social_media_dopamine_productivity.csv")
print(df1.columns.tolist())

print(df1["productivity_self_rating"].describe())
print(df1["productivity_self_rating"].unique())

# Dataset 1 - Social Media/Dopamine
total_cells_1 = df1.size
missing_cells_1 = df1.isnull().sum().sum()
pct_missing_1 = (missing_cells_1 / total_cells_1) * 100

print(f"Dataset 1 - Total cells: {total_cells_1}")
print(f"Dataset 1 - Missing cells: {missing_cells_1}")
print(f"Dataset 1 - Overall missing: {pct_missing_1:.2f}%")
print(f"Duplicate rows: {df1.duplicated().sum()}")

df1['avg_daily_sm_hours'].plot(kind='box')
plt.show()
```
